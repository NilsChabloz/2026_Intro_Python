## Exercise 8: Default arguments and returning two values

Write a function `summarise(values, scale=1.0)` that multiplies every value by `scale` and
returns **two** things: the total and the mean.

1. Call it on `[6.1, 7.4, 5.9, 8.2, 6.8]` with the default scale, unpacking the result into
   `total_celsius` and `mean_celsius`.
2. Call it again with `scale=2.0` passed by keyword, and print only the mean.
3. Call it once more and store the result in a single name. Print its `type` — what did you get?

````python
numbers = [6.1, 7.4, 5.9, 8.2, 6.8]

def summarise(values, scale = 1.0):
   multiplied = [number * scale for number in values]
   total = 0.0
   mean = 0.0
   for value in multiplied:
    total += value
    mean = total/len(multiplied)
   return [total, mean]

# 1.
print(summarise(numbers))

# 2.
print(summarise(numbers, scale=2.0)[1])

# 3.
result = summarise(numbers)
print(type(result))
````

## Exercise 9: A variable number of arguments

Write a function `warmest(*temps_celsius)` (no type hints) that returns the highest value among
any number of positional arguments, using a loop and a running maximum (not `max`). Test it on
three individual readings passed directly, and again by unpacking the list
`[6.1, 7.4, 5.9, 8.2, 6.8]` with `*`.

````python
def warmest(*temps_celsius):
  highest = temps_celsius[0]
  for v in temps_celsius:
    if v > highest:
      highest = v
  return highest

print(warmest(3.4, 5.2, 2.8))

readings = [6.1, 7.4, 5.9, 8.2, 6.8]
print(warmest(*readings))
````

````python
def drop_last(readings):
    readings.pop()
    return readings

original = [6.1, 7.4, 5.9]

print(original)
print(drop_last(original))
print(original) #we deleted the last value of the original list

def drop_last(readings):
    readings_list = readings.copy()
    readings_list.pop()
    return readings_list

original = [6.1, 7.4, 5.9]

print(original)
print(drop_last(original))
print(original)

# Like the mutable-default bug: the function received the caller's list itself
# (not a copy), so .pop() also changed the original.
````

### Exercise 15: Reject impossible temperatures

Write `to_kelvin(temp_celsius)` that raises a `ValueError` when the input is below absolute zero
(−273.15 °C) and otherwise returns the temperature in kelvin. Give it a docstring and a type hint.

Loop over `[20.0, -300.0, 0.0]`, calling it inside a `try`/`except` so that the invalid value is
reported without stopping the loop.

````python
def to_kelvin(temp_celsius: float) -> float:
  ''' Convert a float from celsius to kelvin. It also checks if
  the celsius temperature is not under absolute zero, otherwise it will raise
  an error. '''

  if temp_celsius < -273.15:
    raise ValueError('Temperature is below absolute zero')
  return temp_celsius + 273.15

temp_celsius = [20.0, -300.0, 0.0]
nb_i = 0

for t in temp_celsius:
  nb_i += 1
  try:
    print(to_kelvin(t))
  except ValueError as err:
    print(f'At iteration {nb_i}, skipped t = {t}:', err)
````

**Step 1.** Write a function `fahrenheit_to_celsius(temp_fahrenheit)` that returns the

**Step 4.** Build a second dictionary mapping each month to its **mean** daily maximum
temperature in °C. Print it with each mean rounded to two decimals.

````python
from pathlib import Path

with Path(data_file).open('r', encoding = 'utf-8') as f:
  lines = f.read().strip().split('\n')
records = lines[1:]

counts = {}
converted = 0
skipped = 0
total_temp = {}
mean_temp = {}

for record in records:
  # separate records into three strings: code, date and temp_fahrenheit
  code, date, temp_fahrenheit = record.split(',')

  # tests wether the temperature has been observed or is empty
  if temp_fahrenheit == '':
    # counts the number of date skipped
    skipped += 1
    continue

  # converts fahrenheit to celsius
  temp_celsius = fahrenheit_to_celsius(float(temp_fahrenheit))

  # counts the number of observation converted
  converted += 1

  # splits the date into day, month and year
  day, month, year = date.split('.')

  # counts how many observation there is per month:
  # if there is no observation recorded for a new mounth yet in counts = {},
  # .get(month, 0) will give it a default value of 0. The + 1 adds 1 to the
  # previous number in counts[month] as it is itirating over all days, but checks
  # the months.
  counts[month] = counts.get(month, 0) + 1

  # running total per month
  total_temp[month] = total_temp.get(month, 0) + temp_celsius

# calculates the mean temperature of each month by dividing the total temp per
# month by the number of observations per month
for month in total_temp:
  mean_temp[month] = total_temp[month]/counts[month]

for month in mean_temp:
  print(f'{month}: {mean_temp[month]:.2f}')

# Hint: accumulate a running total per month in the same loop as the counts
# Hint: a second loop over the totals turns each total into a mean
# Hint: reuse your function from step 1 — do not repeat the conversion arithmetic
````

````python
from pathlib import Path

with Path(data_file).open('r', encoding = 'utf-8') as f:
  lines = f.read().strip().split('\n')
records = lines[1:]

counts = {}
converted = 0
skipped = 0
total_temp = {}
mean_temp = {}

for record in records:
  # separate records into three strings: code, date and temp_fahrenheit
  fields = record.split(',')

  if len(fields) != 3:
      skipped += 1
      continue

  code, date, temp_fahrenheit = fields

  # tests wether the temperature has been observed or is empty
  if temp_fahrenheit == '':
    # counts the number of date skipped
    skipped += 1
    continue

  # converts fahrenheit to celsius
  try:
    temp_celsius = fahrenheit_to_celsius(float(temp_fahrenheit))
  except ValueError as err:
    print(f'skipped {fields}: {err}')
    skipped +=1
    continue

  # counts the number of observation converted
  converted += 1

  # splits the date into day, month and year
  day, month, year = date.split('.')

  # counts how many observation there is per month:
  # if there is no observation recorded for a new mounth yet in counts = {},
  # .get(month, 0) will give it a default value of 0. The + 1 adds 1 to the
  # previous number in counts[month] as it is itirating over all days, but checks
  # the months.
  counts[month] = counts.get(month, 0) + 1

  # running total per month
  total_temp[month] = total_temp.get(month, 0) + temp_celsius

# calculates the mean temperature of each month by dividing the total temp per
# month by the number of observations per month
for month in total_temp:
  mean_temp[month] = total_temp[month]/counts[month]

for month in mean_temp:
  print(f'{month}: {mean_temp[month]:.2f}')

# Hint: accumulate a running total per month in the same loop as the counts
# Hint: a second loop over the totals turns each total into a mean
# Hint: reuse your function from step 1 — do not repeat the conversion arithmetic
````

**Q10) Now write a single function that converts to either Earth masses or Jupiter
masses, depending on a keyword argument.**

Call it `to_planet_masses(mass_1e24kg, reference="jupiter")`. Converting Jupiter's own mass should
give 317.923 with `reference="earth"` and exactly 1.0 with `reference="jupiter"`.

With the default argument, one function covers both conversions. Jupiter in Jupiter masses has
to come out as exactly 1.0; any other value means the function divides by the wrong mass.

````python
# Q10: write to_planet_masses here, then test both references on Jupiter
def to_planet_masses(mass_1e24kg, reference="jupiter"):
  if reference =='jupiter':
    planet_masses = mass_1e24kg/planet_masses_1e24kg[reference]
  elif reference == 'earth':
    planet_masses = mass_1e24kg/planet_masses_1e24kg[reference]
  return planet_masses

print(to_planet_masses(1898, reference='jupiter'))
print(to_planet_masses(1898, reference='earth'))
````
